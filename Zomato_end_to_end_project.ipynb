```python
ratings=df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index(name='Count').sort_values(by='Count',ascending=False)
```

```python
ratings
```

```python
plt.bar(ratings['Aggregate rating'],ratings['Count'],color='Skyblue')
plt.xlabel('Aggregate rating')
plt.ylabel('Count')
plt.title('Distribution of Aggregate ratings in Zomato')
```
